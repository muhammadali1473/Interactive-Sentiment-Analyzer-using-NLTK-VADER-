```python
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer


nltk.download('vader_lexicon')

class SentimentAnalyzer:
    def __init__(self):
        self.sia = SentimentIntensityAnalyzer()

    def analyze_sentiment(self, text):
        """
        Analyze the sentiment of given text
        Returns: 'Positive', 'Negative', or 'Neutral'
        """
        sentiment_scores = self.sia.polarity_scores(text)
        compound = sentiment_scores['compound']

        if compound >= 0.05:
            return 'Positive'
        elif compound <= -0.05:
            return 'Negative'
        else:
            return 'Neutral'

def main():
    analyzer = SentimentAnalyzer()
    print("🔍 Sentiment Analyzer")
    print("Type a sentence to analyze its sentiment.")
    print("Type 'exit' to quit.\n")

    while True:
        user_input = input("Enter your text: ").strip()
        if user_input.lower() == "exit":
            print("\n👋 Exiting Sentiment Analyzer. Goodbye!")
            break
        if not user_input:
            print("⚠️ Please enter some text.\n")
            continue

        sentiment = analyzer.analyze_sentiment(user_input)
        print(f"➡️ Sentiment: {sentiment}\n")

if __name__ == "__main__":
    main()
```
